```python
import pandas as pd
import joblib
from sentence_transformers import SentenceTransformer
import os

# ==========================
# FILE PATH
# ==========================

excel_file = r"C:\Users\ESR Solutions\Desktop\RentAAA Chatbot\Data\question_answers.xlsx"
model_folder = r"C:\Users\ESR Solutions\Desktop\RentAAA Chatbot\Models"

# ==========================
# LOAD DATASET
# ==========================

print("Loading Excel dataset...")

df = pd.read_excel(excel_file)

questions = df["Question"].astype(str).tolist()
answers = df["Answer"].astype(str).tolist()

print("Total Questions:", len(questions))

# ==========================
# LOAD EMBEDDING MODEL
# ==========================

print("Loading embedding model...")

model = SentenceTransformer("all-MiniLM-L6-v2")

# ==========================
# CREATE EMBEDDINGS
# ==========================

print("Generating embeddings...")

embeddings = model.encode(questions)

# ==========================
# SAVE DATA
# ==========================

joblib.dump(questions, os.path.join(model_folder, "questions1.pkl"))
joblib.dump(answers, os.path.join(model_folder, "answers2.pkl"))
joblib.dump(embeddings, os.path.join(model_folder, "embeddings3.pkl"))

print("Training completed successfully!")
```
